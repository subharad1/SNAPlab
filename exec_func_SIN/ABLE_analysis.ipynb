```python
# IM and Age
model_IM_age = smf.ols('IM ~ age', data=able).fit()
print(model_IM_age.summary())

predictions = model_IM_age.get_prediction(able[['age']])
prediction_summary = predictions.summary_frame(alpha=0.05)  # 95% CI
adj_r_squared = model_IM_age.rsquared_adj
p_value = model_IM_age.f_pvalue

sorted_idx = able['age'].argsort()
sorted_age = able['age'].iloc[sorted_idx]
sorted_mean = prediction_summary['mean'].iloc[sorted_idx]
sorted_ci_lower = prediction_summary['mean_ci_lower'].iloc[sorted_idx]
sorted_ci_upper = prediction_summary['mean_ci_upper'].iloc[sorted_idx]

plt.figure(figsize=(7, 5))
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['pdf.fonttype'] = 42  
plt.rcParams['svg.fonttype'] = 'none'

plt.scatter(able['age'], able['IM'], alpha=0.5)

plt.plot(sorted_age, sorted_mean, color='black')

plt.fill_between(sorted_age, sorted_ci_lower, sorted_ci_upper, color='black', alpha=0.3)

plt.xlabel('Age',fontsize=12)
plt.ylabel("IM score",fontsize=12)
# plt.legend()
plt.text(0.05, 0.9, f'$R^2$: {adj_r_squared:.3f}', transform=plt.gca().transAxes, fontsize=10)
plt.text(0.05, 0.85, f'p-value: {p_value:.5f}', transform=plt.gca().transAxes, fontsize=10)
plt.title('IM score with Age')
# plt.savefig(foldername+'Fig1/IM_age_v2.pdf', format='pdf')
# plt.savefig(foldername+'Fig1/IM_age_v2.png')
plt.show()
```

```python
def add_jitter(values, scale=0.025):
    return values + np.random.uniform(-scale, scale, size=len(values))

colors = ["red", "gray", "#1f77b4"]
colums_data = ['1 back', '2 back', '3 back']

plt.figure(figsize=(7, 5))

spacing_factor = 0.2

for idx, (col, color) in enumerate(zip(["d1", "d2", "d3"], colors)):
    x_position = idx * spacing_factor  # Calculate new x position
    jittered_x = add_jitter(np.full(len(able[col]), x_position))  # Apply jitter around the position
    plt.scatter(jittered_x, able[col], color=color, alpha=0.3, label=colums_data[idx])
    # plt.scatter(x_position, able[col].mean(), color=color, s=100, marker='s', alpha=1, linewidth=1.5)
    plt.errorbar(x_position, means_nback[idx], yerr=error_nback[idx],color=color, ecolor='black', capsize=3, markersize=7, marker='s')


# Set x-ticks to match the adjusted positions
tick_positions = [i * spacing_factor for i in range(len(colums_data))]
plt.xticks(tick_positions, colums_data)
plt.ylabel("N back d'", fontsize=12)
plt.xlabel("N back", fontsize=12)
plt.title("N back performance")
# plt.legend(loc="upper left", bbox_to_anchor=(1, 1))
plt.legend()
# plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
# plt.savefig(foldername+'Fig2/N_back_dprime_mean_v2.pdf', format='pdf')
# plt.savefig(foldername+'Fig2/N_back_dprime_mean_v2.png')
plt.show()
```
